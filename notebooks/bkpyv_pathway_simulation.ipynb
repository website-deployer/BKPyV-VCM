## Pathway Profile Functions

These functions create custom pathway activity profiles based on single-cell data patterns.

```python
def create_custom_pathway_profile(profile_name: str) -> dict:
    """Create custom pathway activity profiles based on single-cell data patterns.

    These profiles represent different infection states observed in single-cell
    transcriptomic studies of BKPyV nephropathy:

    Args:
        profile_name: Name of the profile to create
            - 'healthy': Healthy kidney cell baseline
            - 'low_viral': Low viral replication state
            - 'high_viral': High viral replication state
            - 'ddr_activated': DNA damage response activated
            - 'immune_active': Strong immune response

    Returns:
        Dictionary of pathway activities
    """
    profiles = {
        'healthy': {
            'dna_replication': 0.8,
            'cell_cycle': 0.6,
            'dna_damage_response': 0.3,
            'innate_immune': 0.5,
            'apoptosis': 0.2,
            'mTOR_signaling': 1.0,
            'viral_replication': 0.0,
            'cellular_stress': 0.3,
            'interferon_response': 0.4,
            'nucleotide_synthesis': 0.7,
        },
        'low_viral': {
            'dna_replication': 0.9,
            'cell_cycle': 0.8,
            'dna_damage_response': 0.5,
            'innate_immune': 0.6,
            'apoptosis': 0.3,
            'mTOR_signaling': 1.0,
            'viral_replication': 0.3,
            'cellular_stress': 0.5,
            'interferon_response': 0.6,
            'nucleotide_synthesis': 0.9,
        },
        'high_viral': {
            'dna_replication': 1.5,
            'cell_cycle': 1.4,
            'dna_damage_response': 0.7,
            'innate_immune': 0.3,
            'apoptosis': 0.4,
            'mTOR_signaling': 1.2,
            'viral_replication': 1.2,
            'cellular_stress': 0.8,
            'interferon_response': 0.2,
            'nucleotide_synthesis': 1.3,
        },
        'ddr_activated': {
            'dna_replication': 0.4,
            'cell_cycle': 0.3,
            'dna_damage_response': 1.5,
            'innate_immune': 0.5,
            'apoptosis': 0.7,
            'mTOR_signaling': 0.8,
            'viral_replication': 0.5,
            'cellular_stress': 1.2,
            'interferon_response': 0.5,
            'nucleotide_synthesis': 0.5,
        },
        'immune_active': {
            'dna_replication': 0.6,
            'cell_cycle': 0.5,
            'dna_damage_response': 0.6,
            'innate_immune': 1.5,
            'apoptosis': 0.5,
            'mTOR_signaling': 0.9,
            'viral_replication': 0.2,
            'cellular_stress': 0.6,
            'interferon_response': 1.4,
            'nucleotide_synthesis': 0.6,
        },
    }

    return profiles.get(profile_name, profiles['healthy'])
```
